```python
import pandas as pd

attack_df = pd.read_csv('data/CTU13_Attack_Traffic.csv')
normal_df = pd.read_csv('data/CTU13_Normal_Traffic.csv')

attack_df['Label'] = 1
normal_df['Label'] = 0

df = pd.concat([attack_df, normal_df], ignore_index=True)

print("Merged dataset shape:", df.shape)

# Save merged dataset
df.to_csv('data/CTU13_merged.csv', index=False)
print("Merged dataset saved to data/CTU13_merged.csv ✅")
```
